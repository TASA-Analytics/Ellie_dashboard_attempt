# 8-Tier Emissions Sankey Chart (Global-Ranked)

**Ranking logic:** Sectors are ranked *globally* by total emissions at each tier.
- Tier 1: top 9 globally → rest into Other
- Tier 2: top 5 globally (summed across all T1 parents) → rest into Other
- Tier 3: top 5 globally (summed across all T2 parents) → rest into Other
- Other chains: T1 Other → T2 Other → T3 Other

**Set `FILE_PATH` and `SECTOR_NAME` before running.**


```python
# ─── CONFIGURATION ─────────────────────────────────────────────────────────
FILE_PATH   = "TASA-EFX_rLCA_Abrasive-Material_KOR_2023_v1.0_DEMO.xlsx"
SECTOR_NAME = "Abrasive Material"
YEAR        = "2023"

TOP_N_T1 = 9   # Top N Tier-1 sectors shown globally
TOP_N_T2 = 5   # Top N Tier-2 sectors shown globally (summed across all T1 parents)
TOP_N_T3 = 5   # Top N Tier-3 sectors shown globally (summed across all T2 parents)
# ───────────────────────────────────────────────────────────────────────────
```

```python
import json
import pandas as pd
import numpy as np
from IPython.display import display, HTML

# ── 1. Load top-level totals from Sector Emissions Factors ───────────────────
ef = pd.read_excel(FILE_PATH, sheet_name="Sector Emissions Factors",
                   header=0, dtype=str)
ef.columns = [str(c).strip() for c in ef.columns]

# Filter to the chosen sector row
sector_row = ef[ef.iloc[:, 0].str.strip() == SECTOR_NAME].iloc[0]

# B2 = col index 1 (0-based), C2=2, D2=3, E2=4
total_ei = float(sector_row.iloc[1])
s1_total = float(sector_row.iloc[2])
s2_total = float(sector_row.iloc[3])
s3_total = float(sector_row.iloc[4])

print(f"Sector : {SECTOR_NAME}")
print(f"Total  : {total_ei:.4f}")
print(f"Scope 1: {s1_total:.4f}")
print(f"Scope 2: {s2_total:.4f}")
print(f"Scope 3: {s3_total:.4f}")
```

```python
# ── 2. Load Tier 1 Pathway Analysis ─────────────────────────────────────────
t1_raw = pd.read_excel(FILE_PATH, sheet_name="Tier_1 Pathway Analysis",
                       header=0, dtype=str)
t1_raw.columns = [str(c).strip() for c in t1_raw.columns]

# Filter to sector
sector_col = t1_raw.columns[0]        # "Sector"
t1_sec     = t1_raw[t1_raw[sector_col].str.strip() == SECTOR_NAME].copy()

# Identify column names (flexible – finds by keyword)
def find_col(df, *keywords):
    """Return first column whose name contains all keywords (case-insensitive)."""
    kws = [k.lower() for k in keywords]
    for c in df.columns:
        cl = c.lower()
        if all(k in cl for k in kws):
            return c
    raise KeyError(f"No column matching {keywords} in {list(df.columns)}")

T1_SECTOR_COL = find_col(t1_sec, "tier 1", "input")
T1_TOTAL_COL  = find_col(t1_sec, "tier 1", "total")
T1_S1_COL     = find_col(t1_sec, "tier 1", "scope 1")
T1_S2_COL     = find_col(t1_sec, "tier 1", "scope 2")
T1_S3_COL     = find_col(t1_sec, "tier 1", "scope 3")

for c in [T1_TOTAL_COL, T1_S1_COL, T1_S2_COL, T1_S3_COL]:
    t1_sec[c] = pd.to_numeric(t1_sec[c], errors="coerce")

# De-dup on Tier 1 Input Sector, keep max total if duplicates exist
t1_sec = (t1_sec
          .dropna(subset=[T1_TOTAL_COL])
          .sort_values(T1_TOTAL_COL, ascending=False)
          .drop_duplicates(subset=[T1_SECTOR_COL]))

# Split Other vs non-Other, take top N non-Other, then bucket the rest
is_other_mask = t1_sec[T1_SECTOR_COL].str.strip().str.lower() == "other"
t1_non_other  = t1_sec[~is_other_mask].head(TOP_N_T1)
t1_rest       = t1_sec[~is_other_mask].iloc[TOP_N_T1:]   # sectors beyond top-N
t1_explicit_other = t1_sec[is_other_mask]

# Combine rest + explicit Other rows into a single "Other" bucket
all_other = pd.concat([t1_rest, t1_explicit_other])
if not all_other.empty:
    other_row = pd.DataFrame([{
        T1_SECTOR_COL: "Other",
        T1_TOTAL_COL:  all_other[T1_TOTAL_COL].sum(),
        T1_S1_COL:     all_other[T1_S1_COL].sum(),
        T1_S2_COL:     all_other[T1_S2_COL].sum(),
        T1_S3_COL:     all_other[T1_S3_COL].sum(),
    }])
    t1_final = pd.concat([t1_non_other, other_row]).reset_index(drop=True)
else:
    t1_final = t1_non_other.reset_index(drop=True)

print(f"Tier 1 sectors shown: {len(t1_final)}")
print(f"Tier 1 sectors sum: {t1_final[T1_TOTAL_COL].sum()}")
t1_final[[T1_SECTOR_COL, T1_TOTAL_COL]].head(12)
```

```python
# ── CHECK: Tier 1 reconciliation ────────────────────────────────────────────
print("=" * 70)
print("TIER 1 RECONCILIATION")
print("=" * 70)

# Reference: what the source sheet says the sector total is
ref_s1 = s1_total
ref_s2 = s2_total
ref_s3 = s3_total
ref_total = total_ei

# What our t1_final sums to
our_total = t1_final[T1_TOTAL_COL].sum()
our_s1    = t1_final[T1_S1_COL].sum()
our_s2    = t1_final[T1_S2_COL].sum()
our_s3    = t1_final[T1_S3_COL].sum()

print(f"{'Metric':<20} {'Source':>12} {'T1 sum':>12} {'Diff':>12} {'OK?':>6}")
print("-" * 64)
for name, ref, our in [("Total",  ref_total, our_total),
                        ("Scope 1", ref_s1,   our_s1),
                        ("Scope 2", ref_s2,   our_s2),
                        ("Scope 3", ref_s3,   our_s3)]:
    diff = our - ref
    ok   = "✓" if abs(diff) < 0.001 else "✗ ← MISMATCH"
    print(f"{name:<20} {ref:>12.4f} {our:>12.4f} {diff:>12.4f} {ok:>6}")

print()
print("T1 final rows:")
print(t1_final[[T1_SECTOR_COL, T1_TOTAL_COL, T1_S1_COL, T1_S2_COL, T1_S3_COL]].to_string(index=False))

# Also check: does T1 raw have any duplicate sector rows we dropped?
raw_dupes = t1_sec[t1_sec.duplicated(subset=[T1_SECTOR_COL], keep=False)]
if not raw_dupes.empty:
    print(f"\n⚠ RAW T1 DUPLICATES FOUND ({len(raw_dupes)} rows):")
    print(raw_dupes[[T1_SECTOR_COL, T1_TOTAL_COL]].to_string(index=False))
else:
    print("\n✓ No duplicate T1 sector rows found")
print("=" * 70)
```

```python
# ── 3. Load Tier 2 Pathway Analysis (global ranking, total = S1+S2+S3) ──────
t2_raw = pd.read_excel(FILE_PATH, sheet_name="Tier_2 Pathway Analysis",
                       header=0, dtype=str)
t2_raw.columns = [str(c).strip() for c in t2_raw.columns]

t2_sec = t2_raw[t2_raw.iloc[:, 0].str.strip() == SECTOR_NAME].copy()

T2_T1_COL     = find_col(t2_sec, "tier 2", "sector")   if any("sector" in c.lower() and "tier 2" in c.lower() for c in t2_sec.columns) else find_col(t2_sec, "tier 1", "input")
T2_T1_COL     = find_col(t2_sec, "tier 1", "input")
T2_SECTOR_COL = find_col(t2_sec, "tier 2", "input")
T2_TOTAL_COL  = find_col(t2_sec, "tier 2", "total")
T2_S1_COL     = find_col(t2_sec, "tier 2", "scope 1")
T2_S2_COL     = find_col(t2_sec, "tier 2", "scope 2")
T2_S3_COL     = find_col(t2_sec, "tier 2", "scope 3")

for c in [T2_TOTAL_COL, T2_S1_COL, T2_S2_COL, T2_S3_COL]:
    t2_sec[c] = pd.to_numeric(t2_sec[c], errors="coerce")
t2_sec = t2_sec.dropna(subset=[T2_S1_COL, T2_S2_COL, T2_S3_COL])

# Derive contribution total as S1+S2+S3 (do NOT use the Total column)
t2_sec["_contrib"] = t2_sec[T2_S1_COL] + t2_sec[T2_S2_COL] + t2_sec[T2_S3_COL]

# ── Global ranking: sum each T2 sector's S3 contribution across ALL T1 parents
# We rank by S3 because T2 explains Scope 3 of T1 — S3 is the relevant flow.
# Using S3 for ranking avoids the double-counting that the Total column caused.
t2_is_other = t2_sec[T2_SECTOR_COL].str.strip().str.lower() == "other"
t2_named    = t2_sec[~t2_is_other].copy()

t2_global_s3 = (
    t2_named.groupby(t2_named[T2_SECTOR_COL].str.strip())["_contrib"]
    .sum().sort_values(ascending=False)
)
top_t2_names = set(t2_global_s3.head(TOP_N_T2).index)
print(f"Global top-{TOP_N_T2} Tier-2 sectors (ranked by summed S1+S2+S3):")
for nm in sorted(top_t2_names):
    print(f"  {nm}: {t2_global_s3[nm]:.4f}")

# ── Build t2_by_t1 ────────────────────────────────────────────────────────────
# Each T1 parent's T2 rows represent that parent's Scope 3 breakdown.
# Values: use S1, S2, S3 directly from the sheet; total = S1+S2+S3.
t2_by_t1 = {}

for _, t1_row in t1_final.iterrows():
    t1_name = str(t1_row[T1_SECTOR_COL]).strip()

    if t1_name == "Other":
        # T1-Other → T2-Other passthrough using T1's own S3 value
        t2_by_t1[t1_name] = [{
            "label": "Other",
            "total": float(t1_row[T1_S3_COL]),
            "s1": 0.0, "s2": 0.0, "s3": float(t1_row[T1_S3_COL])
        }]
        continue

    subset = t2_sec[t2_sec[T2_T1_COL].str.strip() == t1_name].copy()
    # No dedup needed — each (T1, T2) pair should have exactly one row;
    # if duplicates exist, sum them (same label under same parent = same thing measured twice)
    subset["_label"] = subset[T2_SECTOR_COL].str.strip()
    subset = subset.groupby("_label", as_index=False).agg({
        T2_S1_COL: "sum", T2_S2_COL: "sum", T2_S3_COL: "sum", "_contrib": "sum"
    })
    subset[T2_SECTOR_COL] = subset["_label"]

    rows = []
    other_s1 = other_s2 = other_s3 = 0.0

    for _, r in subset.iterrows():
        lbl = str(r[T2_SECTOR_COL]).strip()
        s1  = float(r[T2_S1_COL])
        s2  = float(r[T2_S2_COL])
        s3  = float(r[T2_S3_COL])
        tot = s1 + s2 + s3   # always derive

        if lbl.lower() == "other" or lbl not in top_t2_names:
            other_s1 += s1; other_s2 += s2; other_s3 += s3
        else:
            rows.append({"label": lbl, "total": tot, "s1": s1, "s2": s2, "s3": s3})

    other_tot = other_s1 + other_s2 + other_s3
    if other_tot > 0:
        rows.append({"label": "Other", "total": other_tot,
                     "s1": other_s1, "s2": other_s2, "s3": other_s3})

    if rows:
        t2_by_t1[t1_name] = rows

# Verify: each T1's T2 rows should sum to that T1's S3
print("\nT2 S3 sum vs T1 S3 reference:")
for _, t1_row in t1_final.iterrows():
    t1_name = str(t1_row[T1_SECTOR_COL]).strip()
    t1_s3   = float(t1_row[T1_S3_COL])
    our_s3  = sum(r["s3"] for r in t2_by_t1.get(t1_name, []))
    diff    = our_s3 - t1_s3
    flag    = "" if abs(diff) < 0.01 else f"  ✗ DIFF={diff:.4f}"
    print(f"  {t1_name:<40} T1_S3={t1_s3:.4f}  T2_sum={our_s3:.4f}{flag}")
```

```python
# ── DIAGNOSTIC: Understand T2 sheet structure & double-counting ──────────────
print("=" * 70)
print("DIAGNOSTIC: T2 SHEET STRUCTURE")
print("=" * 70)

# Q1: Does the same T2 sector name appear under multiple T1 parents?
t2_counts = t2_sec.groupby(t2_sec[T2_SECTOR_COL].str.strip())[T2_T1_COL].nunique()
multi_parent = t2_counts[t2_counts > 1].sort_values(ascending=False)
print(f"\nT2 sectors appearing under >1 T1 parent: {len(multi_parent)}")
if not multi_parent.empty:
    print(multi_parent.head(10).to_string())

# Q2: For a T2 sector that appears under multiple parents, are the values the same or different?
print("\n── Sample: same T2 sector across different T1 parents ──")
if not multi_parent.empty:
    sample_t2 = multi_parent.index[0]
    rows = t2_sec[t2_sec[T2_SECTOR_COL].str.strip() == sample_t2][[T2_T1_COL, T2_SECTOR_COL, T2_TOTAL_COL, T2_S1_COL, T2_S2_COL, T2_S3_COL]]
    print(f"T2 sector: '{sample_t2}'")
    print(rows.to_string(index=False))
    print(f"Sum across all parents: total={rows[T2_TOTAL_COL].sum():.4f}  S3={rows[T2_S3_COL].sum():.4f}")
    print(f"Are all values identical? {rows[T2_TOTAL_COL].nunique() == 1}")

# Q3: For each T1 parent, do its T2 rows sum to that parent's S3?
print("\n── Do T2 rows sum to T1 S3 (per parent)? ──")
print(f"  {'T1 Sector':<40} {'T1 S3':>10} {'T2 rows sum':>12} {'Match':>8}")
for _, t1_row in t1_final.iterrows():
    t1_name = str(t1_row[T1_SECTOR_COL]).strip()
    if t1_name == "Other":
        continue
    t1_s3 = float(t1_row[T1_S3_COL])
    subset = t2_sec[t2_sec[T2_T1_COL].str.strip() == t1_name]
    t2_s3_sum = subset[T2_S3_COL].sum()
    match = "✓" if abs(t2_s3_sum - t1_s3) < 0.01 else "✗ DIFF={:.4f}".format(t2_s3_sum - t1_s3)
    print(f"  {t1_name:<40} {t1_s3:>10.4f} {t2_s3_sum:>12.4f} {match:>8}")

# Q4: What IS the T2 total column — is it per-parent contribution or T2's own total?
print("\n── Key question: is T2 Total = T2 own total, or T2's contribution to this T1? ──")
if not multi_parent.empty:
    sample_t2 = multi_parent.index[0]
    rows = t2_sec[t2_sec[T2_SECTOR_COL].str.strip() == sample_t2]
    # If it's the T2's own total, all rows should have same value
    # If it's the contribution, values should vary by parent and sum to something meaningful
    vals = rows[T2_TOTAL_COL].values
    print(f"  T2 sector '{sample_t2}' total values across parents: {[round(v,4) for v in vals]}")
    print(f"  All same? {len(set(round(v,6) for v in vals)) == 1}")
    print(f"  Their sum: {sum(vals):.4f}")
    # Compare to T1 total for same sector if it exists
    t1_match = t1_sec[t1_sec[T1_SECTOR_COL].str.strip() == sample_t2] if sample_t2 in t1_sec[T1_SECTOR_COL].str.strip().values else None
    if t1_match is not None and not t1_match.empty:
        print(f"  T1 total for same name '{sample_t2}': {float(t1_match.iloc[0][T1_TOTAL_COL]):.4f}")

print("=" * 70)
```

```python
# ── CHECK: Tier 2 reconciliation ────────────────────────────────────────────
print("=" * 70)
print("TIER 2 RECONCILIATION")
print("=" * 70)

# For each T1 sector (excluding T1-Other), the T2 rows under it should sum
# to the T1 sector's own S3 value (since T2 explains Scope 3 of T1)
print(f"\n{'T1 Sector':<35} {'T1 S3 (ref)':>12} {'T2 sum total':>13} {'T2 sum S1':>10} {'T2 sum S2':>10} {'T2 sum S3':>10} {'S3 diff':>9}")
print("-" * 105)

t2_s3_errors = []
for _, t1_row in t1_final.iterrows():
    t1_name = str(t1_row[T1_SECTOR_COL]).strip()
    t1_s3_ref = float(t1_row[T1_S3_COL])
    
    if t1_name == "Other":
        # Show what we passed through vs what the raw T2 rows actually sum to
        raw_under_other = t2_sec[~t2_sec[T2_T1_COL].str.strip().isin(
            [str(r[T1_SECTOR_COL]).strip() for _, r in t1_final.iterrows() if str(r[T1_SECTOR_COL]).strip() != "Other"]
        )]
        raw_s3 = raw_under_other[T2_S3_COL].sum() if not raw_under_other.empty else 0
        our_rows = t2_by_t1.get(t1_name, [])
        our_s3 = sum(r["s3"] for r in our_rows)
        diff = our_s3 - t1_s3_ref
        flag = " ← PASSTHROUGH (raw T2 S3 would be {:.4f})".format(raw_s3)
        print(f"{'(Other bucket)':<35} {t1_s3_ref:>12.4f} {'—':>13} {'—':>10} {'—':>10} {our_s3:>10.4f} {diff:>9.4f}{flag}")
        continue
    
    # Raw T2 rows under this T1
    raw_subset = t2_sec[t2_sec[T2_T1_COL].str.strip() == t1_name]
    raw_total  = raw_subset[T2_TOTAL_COL].sum()
    raw_s1     = raw_subset[T2_S1_COL].sum()
    raw_s2     = raw_subset[T2_S2_COL].sum()
    raw_s3     = raw_subset[T2_S3_COL].sum()
    
    # After dedup
    deduped    = raw_subset.drop_duplicates(subset=[T2_SECTOR_COL])
    ded_total  = deduped[T2_TOTAL_COL].sum()
    ded_s3     = deduped[T2_S3_COL].sum()
    
    # What our t2_by_t1 has
    our_rows   = t2_by_t1.get(t1_name, [])
    our_total  = sum(r["total"] for r in our_rows)
    our_s1     = sum(r["s1"]    for r in our_rows)
    our_s2     = sum(r["s2"]    for r in our_rows)
    our_s3     = sum(r["s3"]    for r in our_rows)
    
    diff_s3 = our_s3 - t1_s3_ref
    flag = " ← MISMATCH" if abs(diff_s3) > 0.01 else ""
    print(f"{t1_name:<35} {t1_s3_ref:>12.4f} {our_total:>13.4f} {our_s1:>10.4f} {our_s2:>10.4f} {our_s3:>10.4f} {diff_s3:>9.4f}{flag}")
    
    if abs(raw_total - ded_total) > 0.001:
        print(f"  ⚠ DEDUP dropped rows! Raw sum={raw_total:.4f}, after dedup={ded_total:.4f}, dropped={raw_total-ded_total:.4f}")
        dropped = raw_subset[raw_subset.duplicated(subset=[T2_SECTOR_COL], keep='first')]
        print(f"  Dropped sectors: {list(dropped[T2_SECTOR_COL].values)}")
    
    if abs(diff_s3) > 0.01:
        t2_s3_errors.append(t1_name)
        # Show what labels are in our_rows vs raw
        our_labels = [r["label"] for r in our_rows]
        raw_labels = sorted(deduped[T2_SECTOR_COL].str.strip().tolist())
        print(f"  Our labels : {our_labels}")
        print(f"  Raw labels : {raw_labels}")
        print(f"  raw_s3={raw_s3:.4f}  ded_s3={ded_s3:.4f}  our_s3={our_s3:.4f}  ref={t1_s3_ref:.4f}")

print()
if t2_s3_errors:
    print(f"✗ T2 S3 mismatches on: {t2_s3_errors}")
else:
    print("✓ All T2 S3 sums match T1 S3 references")

# Overall T2 sums vs expected
exp_s3 = t1_final[T1_S3_COL].sum()
got_s3 = sum(r["s3"] for rows in t2_by_t1.values() for r in rows)
print(f"\nOverall T2 S3 sum: got={got_s3:.4f}  expected (T1 S3 total)={exp_s3:.4f}  diff={got_s3-exp_s3:.4f}")
print("=" * 70)
```

```python
# ── 4. Load Tier 3 Pathway Analysis (global ranking, total = S1+S2+S3) ──────
t3_raw = pd.read_excel(FILE_PATH, sheet_name="Tier_3 Pathway Analysis",
                       header=0, dtype=str)
t3_raw.columns = [str(c).strip() for c in t3_raw.columns]

t3_sec = t3_raw[t3_raw.iloc[:, 0].str.strip() == SECTOR_NAME].copy()

T3_T1_COL     = find_col(t3_sec, "tier 1", "input")
T3_T2_COL     = find_col(t3_sec, "tier 2", "input")
T3_SECTOR_COL = find_col(t3_sec, "tier 3", "input")
T3_TOTAL_COL  = find_col(t3_sec, "tier 3", "total")
T3_S1_COL     = find_col(t3_sec, "tier 3", "scope 1")
T3_S2_COL     = find_col(t3_sec, "tier 3", "scope 2")
T3_S3_COL     = find_col(t3_sec, "tier 3", "scope 3")

for c in [T3_TOTAL_COL, T3_S1_COL, T3_S2_COL, T3_S3_COL]:
    t3_sec[c] = pd.to_numeric(t3_sec[c], errors="coerce")
t3_sec = t3_sec.dropna(subset=[T3_S1_COL, T3_S2_COL, T3_S3_COL])

# Derive contribution total as S1+S2+S3
t3_sec["_contrib"] = t3_sec[T3_S1_COL] + t3_sec[T3_S2_COL] + t3_sec[T3_S3_COL]

# ── Global ranking by summed S3 contribution across all (T1,T2) parents ───────
t3_is_other = t3_sec[T3_SECTOR_COL].str.strip().str.lower() == "other"
t3_named    = t3_sec[~t3_is_other].copy()

t3_global_s3 = (
    t3_named.groupby(t3_named[T3_SECTOR_COL].str.strip())["_contrib"]
    .sum().sort_values(ascending=False)
)
top_t3_names = set(t3_global_s3.head(TOP_N_T3).index)
print(f"Global top-{TOP_N_T3} Tier-3 sectors (ranked by summed S1+S2+S3):")
for nm in sorted(top_t3_names):
    print(f"  {nm}: {t3_global_s3[nm]:.4f}")

# ── Build t3_by_t2 ────────────────────────────────────────────────────────────
t3_by_t2 = {}

for t1_name, t2_rows in t2_by_t1.items():
    for t2_row in t2_rows:
        t2_name = t2_row["label"]
        key = (t1_name, t2_name)

        if t2_name == "Other" or t1_name == "Other":
            t3_by_t2[key] = [{
                "label": "Other",
                "total": t2_row["s3"],
                "s1": 0.0, "s2": 0.0, "s3": t2_row["s3"]
            }]
            continue

        subset = t3_sec[
            (t3_sec[T3_T1_COL].str.strip() == t1_name) &
            (t3_sec[T3_T2_COL].str.strip() == t2_name)
        ].copy()

        # Sum any duplicate (T1,T2,T3) rows
        subset["_label"] = subset[T3_SECTOR_COL].str.strip()
        subset = subset.groupby("_label", as_index=False).agg({
            T3_S1_COL: "sum", T3_S2_COL: "sum", T3_S3_COL: "sum", "_contrib": "sum"
        })
        subset[T3_SECTOR_COL] = subset["_label"]

        rows = []
        other_s1 = other_s2 = other_s3 = 0.0

        for _, r in subset.iterrows():
            lbl = str(r[T3_SECTOR_COL]).strip()
            s1  = float(r[T3_S1_COL])
            s2  = float(r[T3_S2_COL])
            s3  = float(r[T3_S3_COL])
            tot = s1 + s2 + s3

            if lbl.lower() == "other" or lbl not in top_t3_names:
                other_s1 += s1; other_s2 += s2; other_s3 += s3
            else:
                rows.append({"label": lbl, "total": tot, "s1": s1, "s2": s2, "s3": s3})

        other_tot = other_s1 + other_s2 + other_s3
        if other_tot > 0:
            rows.append({"label": "Other", "total": other_tot,
                         "s1": other_s1, "s2": other_s2, "s3": other_s3})

        if rows:
            t3_by_t2[key] = rows

print(f"\nTier-3 groups loaded: {len(t3_by_t2)}")

# Verify: each T2's T3 rows should sum to that T2's S3
print("\nT3 S3 sum vs T2 S3 reference (first 10 pairs):")
count = 0
for t1_name, t2_rows in t2_by_t1.items():
    for t2_row in t2_rows:
        t2_name  = t2_row["label"]
        t2_s3    = t2_row["s3"]
        key      = (t1_name, t2_name)
        our_s3   = sum(r["s3"] for r in t3_by_t2.get(key, []))
        diff     = our_s3 - t2_s3
        flag     = "" if abs(diff) < 0.01 else f"  ✗ DIFF={diff:.4f}"
        print(f"  {t1_name} → {t2_name:<35} T2_S3={t2_s3:.4f}  T3_sum={our_s3:.4f}{flag}")
        count += 1
        if count >= 15:
            print(f"  ... ({len(t3_by_t2)-count} more pairs)")
            break
    if count >= 15:
        break
```

```python
# ── CHECK: Tier 3 reconciliation ────────────────────────────────────────────
print("=" * 70)
print("TIER 3 RECONCILIATION")
print("=" * 70)

# For each (T1, T2) pair, T3 rows should sum to T2's S3 for that pair
t3_s3_errors = []

print(f"\n{'T1 → T2':<55} {'T2 S3 ref':>10} {'T3 sum':>10} {'Diff':>8}")
print("-" * 88)

for t1_name, t2_rows in t2_by_t1.items():
    for t2_row in t2_rows:
        t2_name   = t2_row["label"]
        t2_s3_ref = t2_row["s3"]
        key       = (t1_name, t2_name)
        our_rows  = t3_by_t2.get(key, [])
        our_s3    = sum(r["s3"] for r in our_rows)
        diff      = our_s3 - t2_s3_ref
        flag      = " ← MISMATCH" if abs(diff) > 0.01 else ""
        label     = f"{t1_name} → {t2_name}"
        print(f"{label:<55} {t2_s3_ref:>10.4f} {our_s3:>10.4f} {diff:>8.4f}{flag}")

        if abs(diff) > 0.01:
            t3_s3_errors.append(key)
            # Investigate: what does the raw T3 sheet have for this pair?
            if t2_name != "Other" and t1_name != "Other":
                raw = t3_sec[
                    (t3_sec[T3_T1_COL].str.strip() == t1_name) &
                    (t3_sec[T3_T2_COL].str.strip() == t2_name)
                ]
                raw_s3 = raw[T3_S3_COL].sum()
                raw_n  = len(raw)
                print(f"  Raw T3 rows for this pair: {raw_n}, raw S3 sum={raw_s3:.4f}")
                if raw_n == 0:
                    # Check if name exists with different casing/spacing
                    t1_matches = t3_sec[T3_T1_COL].str.strip().unique()
                    t2_matches = t3_sec[T3_T2_COL].str.strip().unique()
                    close_t1 = [x for x in t1_matches if t1_name.lower() in x.lower() or x.lower() in t1_name.lower()]
                    close_t2 = [x for x in t2_matches if t2_name.lower() in x.lower() or x.lower() in t2_name.lower()]
                    print(f"  Similar T1 names in T3 sheet: {close_t1[:5]}")
                    print(f"  Similar T2 names in T3 sheet: {close_t2[:5]}")

print()
if t3_s3_errors:
    print(f"✗ T3 S3 mismatches on {len(t3_s3_errors)} pairs: {t3_s3_errors[:5]}")
else:
    print("✓ All T3 S3 sums match T2 S3 references")

# Overall T3 S3 vs T2 S3
exp_s3 = sum(r["s3"] for rows in t2_by_t1.values() for r in rows)
got_s3 = sum(r["s3"] for rows in t3_by_t2.values() for r in rows)
print(f"\nOverall T3 S3 sum: got={got_s3:.4f}  expected (T2 S3 total)={exp_s3:.4f}  diff={got_s3-exp_s3:.4f}")

# Check for T2 name mismatches between t2_by_t1 and T3 sheet
print("\n── T2 label match check (are T2 names in t2_by_t1 findable in T3 sheet?) ──")
t3_t2_names_in_sheet = set(t3_sec[T3_T2_COL].str.strip().unique())
t3_t1_names_in_sheet = set(t3_sec[T3_T1_COL].str.strip().unique())
for t1_name, t2_rows in list(t2_by_t1.items())[:5]:
    for t2_row in t2_rows:
        t2_name = t2_row["label"]
        if t2_name == "Other": continue
        t1_ok = t1_name in t3_t1_names_in_sheet
        t2_ok = t2_name in t3_t2_names_in_sheet
        if not t1_ok or not t2_ok:
            print(f"  ✗ '{t1_name}' → '{t2_name}'  T1_in_sheet={t1_ok}  T2_in_sheet={t2_ok}")
        else:
            print(f"  ✓ '{t1_name}' → '{t2_name}'")
print("=" * 70)
```

```python
# ── 5. Assemble JSON payload ─────────────────────────────────────────────────

def pct(val, base):
    return round(val / base * 100, 2) if base else 0.0

scopes = [
    {"label": "Scope 1", "val": round(s1_total, 4), "pct": pct(s1_total, total_ei), "color": "#3A86C8"},
    {"label": "Scope 2", "val": round(s2_total, 4), "pct": pct(s2_total, total_ei), "color": "#2BAE96"},
    {"label": "Scope 3", "val": round(s3_total, 4), "pct": pct(s3_total, total_ei), "color": "#1B4F8A"},
]

tier1_data = []
for _, row in t1_final.iterrows():
    t_total=float(row[T1_TOTAL_COL]); t_s1=float(row[T1_S1_COL])
    t_s2=float(row[T1_S2_COL]); t_s3=float(row[T1_S3_COL])
    tier1_data.append({
        "label": str(row[T1_SECTOR_COL]).strip(), "val": round(t_total,4),
        "pct": pct(t_total,total_ei),
        "s1_val": round(t_s1,4), "s1": pct(t_s1,total_ei),
        "s2_val": round(t_s2,4), "s2": pct(t_s2,total_ei),
        "s3_val": round(t_s3,4), "s3": pct(t_s3,total_ei),
    })

# tier2_data: keyed by t1_label; each entry is a list of sector rows
# Each sector row now includes "label" which may appear under multiple T1 parents
tier2_data = {}
for t1_label, rows in t2_by_t1.items():
    tier2_data[t1_label] = [{
        "label": r["label"], "val": round(r["total"],4), "pct": pct(r["total"],total_ei),
        "s1_val": round(r["s1"],4), "s1": pct(r["s1"],total_ei),
        "s2_val": round(r["s2"],4), "s2": pct(r["s2"],total_ei),
        "s3_val": round(r["s3"],4), "s3": pct(r["s3"],total_ei),
    } for r in rows]

# tier3_data: keyed by "t1|||t2"
tier3_data = {}
for (t1_label, t2_label), rows in t3_by_t2.items():
    key = t1_label + "|||" + t2_label
    tier3_data[key] = [{
        "label": r["label"], "val": round(r["total"],4), "pct": pct(r["total"],total_ei),
        "s1_val": round(r["s1"],4), "s1": pct(r["s1"],total_ei),
        "s2_val": round(r["s2"],4), "s2": pct(r["s2"],total_ei),
        "s3_val": round(r["s3"],4), "s3": pct(r["s3"],total_ei),
    } for r in rows]

payload = json.dumps({
    "sector": SECTOR_NAME, "total_ei": round(total_ei,4),
    "scopes": scopes, "tier1": tier1_data,
    "tier2": tier2_data, "tier3": tier3_data,
})

print("Payload assembled:")
print(f"  Tier1 sectors: {len(tier1_data)}")
print(f"  Tier2 groups (by T1): {len(tier2_data)}, total rows: {sum(len(v) for v in tier2_data.values())}")
print(f"  Tier3 groups (by T1|T2): {len(tier3_data)}, total rows: {sum(len(v) for v in tier3_data.values())}")
# Show unique T2 sector names to confirm global top-N is working
t2_labels = sorted(set(r["label"] for rows in tier2_data.values() for r in rows))
t3_labels = sorted(set(r["label"] for rows in tier3_data.values() for r in rows))
print(f"  Unique T2 labels: {t2_labels}")
print(f"  Unique T3 labels: {t3_labels}")
```
